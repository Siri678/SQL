```python
import re

def is_valid_expression(expression):
    # This allows only digits, + - * / . ( ) and space
    return bool(re.fullmatch(r"[0-9+\-*/(). ]+", expression))

def calculator():
    print("🧠 Welcome to the BODMAS Calculator!")
    print("💡 Enter a math expression like 2 + 3 * (4 - 1)")
    print("✅ Allowed: Numbers, +, -, *, /, (, ), and space")
    print("👉 Type 'exit' to quit anytime.\n")

    while True:
        user_input = input("🔢 Enter your expression: ").strip()

        if user_input.lower() == 'exit':
            print("👋 Goodbye! Thanks for using the calculator.")
            break

        if not is_valid_expression(user_input):
            print("🚫 Please do NOT enter alphabets or special characters like @, #, $, etc.")
            print("❌ Try again with only numbers and math symbols.\n")
            continue

        try:
            result = eval(user_input)
            print(f"✅ Result: {result}\n")
        except ZeroDivisionError:
            print("❗ Error: Division by zero is not allowed.\n")
        except Exception:
            print("⚠️ Something went wrong. Check your expression.\n")

# Run the calculator
calculator()
```
